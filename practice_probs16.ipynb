```python
#  Write a Python program to divide two numbers and handle division by zero.

a = 50
b = 0
try:
    result = a / b
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero")
```

```python
# Write a program that accepts a number from the user and handles invalid input.

try:
    num = int(input("Enter number: "))
    print("Square:", num * num)
except ValueError:
    print("Invalid input. Please enter an integer")
```

```python
#  Write a program that handles ValueError and ZeroDivisionError separately.

try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print(a / b)
except ValueError:
    print("Please enter numbers only")
except ZeroDivisionError:
    print("b cannot be zero")
```

```python
#  Write an example using try, except, else, and finally.

try:
    num = int(input("Enter number: "))
except ValueError:
    print("Invalid input")
else:
    print("Valid number:", num)
finally:
    print("Execution finished")
```

```python
# Write a Python program to handle list index out of range error.

numbers = [10, 20, 30]
try:
    print(numbers[5])
except IndexError:
    print("Index is out of range")
```

```python
#  Write a program to handle missing dictionary key error.

student = {"name": "Amit", "marks": 85}
try:
    print(student["city"])
except KeyError:
    print("Key not found in dictionary")
```

```python
# Write a program that raises an exception if the number is negative.

num = -5
try:
    if num < 0:
        raise ValueError("Negative numbers are not allowed")
    print("Number is valid")
except ValueError as e:
    print(e)
```

```python
# Create a custom exception for age below 18.

class InvalidAgeError(Exception):
    pass
age = 15
try:
    if age < 18:
        raise InvalidAgeError("Age must be 18 or above")
    print("Eligible")
except InvalidAgeError as e:
    print(e)
```

```python
#  Write a program to open a file safely using exception handling.

try:
    file = open("report.txt", "r")
    print(file.read())
    file.close()
except FileNotFoundError:
    print("File not found")
```

```python
#  Write a program showing how finally is used for cleanup.

try:
    print("Opening database connection")
    result = 10 / 0
except ZeroDivisionError:
    print("Error occurred")
finally:
    print("Closing database connection")
```

```python
#  Create a function safe_divide(a, b) that handles division by zero.

def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Division by zero is not allowed"
print(safe_divide(12, 3))
print(safe_divide(12, 0))
```

```python
# Write a program to handle TypeError when adding incompatible data types.

try:
    result = "Python" + 10
    print(result)
except TypeError:
    print("Cannot add string and integer directly")
```
